```python
# Constants / Settings

RESULTS_DIR1 = "../RESULTS3"


RESULTS_DIR2 = "../RESULTS_OLD"
PLANNERS = ['pr2']

TIME_LIMIT = 3600
```

```python
def results_for_planner(planner, results_dir):
    if planner == '':
        return glob.glob(f'{results_dir}/*')
    else:
        return glob.glob(f'{results_dir}/{planner}.*')
    
def filter_for_domain(results, domain):
    return [res for res in results if f'.{domain}__' in res]

def get_domain_results(results, domain):
    return [process_pr2(res) for res in filter_for_domain(results, domain)]

def extract_instance(res_file):
    return res_file.split('.')[-2]

def file_contains(fn, target):
    with open(fn, 'r') as f:
        content = f.read()
    return target in content

# Methods to extract data

NO_SOL = {
    'solved': False,
    'time': None,
    'search-time': None,
    'size': None
}

def process_pr2(res):
    if not file_contains(res, 'Strong cyclic solution found.'):
        # print(f"PR2 Failed: {res}")
        return NO_SOL
    with open(res, 'r') as f:
        contents = f.read()
    return {
        'solved': True,
        'time': sum([60 * int(m) + float(s) for m, s in re.findall("user\t(\d+)m(\d+.\d+)s", contents)]),
        'search-time': float(re.search(r'Time taken: (\d+.\d+) sec', contents).group(1)),
        'size': int(re.search(r'Solution Size: (\d+)', contents).group(1))
    }

def count_solved(results):
    return sum([1 if x['solved'] else 0 for x in results])

from statistics import mean

def average_search_time(results):
    try:
        return mean([x['search-time'] for x in results if x['search-time'] is not None])
    except:
        return None
    
def average_size(results):
    try:
        return mean([x['size'] for x in results if x['size'] is not None])
    except:
        return None
```

```python
new_results_files = results_for_planner("pr2", RESULTS_DIR1)
old_results_files = results_for_planner("pr2", RESULTS_DIR2)
```

```python
blocksworld_results_new = get_domain_results(new_results_files, "blocksworld-new")
blocksworld_results_old = get_domain_results(old_results_files, "blocksworld-new")

print(blocksworld_results_old)
print(blocksworld_results_new)
```

```python
domains = ["acrobatics",
           "beam-walk",
           "blocksworld-new",
           "chain-of-rooms",
           "doors",
           "earth-observation",
           "elevators",
           "faults-new",
           "first-responders-new",
           "forest-new",
           "islands",
           "miner",
           "tidyup-mdp_tidyip_inst_mdp",
           "tireworld",
           "tireworld-truck",
           "triangle-tireworld",
           "zenotravel"
           ]

for dom in domains:
    print("-------------------------")
    print(dom)
    print("-------------------------")
    print()
    print(f"old")
    print(f" # solved: {count_solved(get_domain_results(old_results_files, dom))}")
    print(f" mean search time: {average_search_time(get_domain_results(old_results_files, dom))}")
    print(f" mean sol size: {average_size(get_domain_results(old_results_files, dom))}")
    print()
    print(f"new")
    print(f" # solved: {count_solved(get_domain_results(new_results_files, dom))}")
    print(f" mean search time: {average_search_time(get_domain_results(new_results_files, dom))}")
    print(f" mean sol size: {average_size(get_domain_results(new_results_files, dom))}")
    print()
```
